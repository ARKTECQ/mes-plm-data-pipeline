```python
machine_logs = []

for i in range(10000):

    product = random.choice(product_df["product_id"])

    status = random.choices(
        machine_status,
        weights=[80, 15, 5],
        k=1
    )[0]

    if status == "Running":
        output_qty = random.randint(40, 120)
        defect_qty = random.randint(0, 5)
    elif status == "Idle":
        output_qty = 0
        defect_qty = 0
    else:
        output_qty = random.randint(0, 10)
        defect_qty = random.randint(0, 2)

    row = {
        "machine_id": random.choice(machine_ids),
        "timestamp": fake.date_time_between(
            start_date="-90d",
            end_date="now"
        ),
        "product_id": product,
        "shift": random.choice(shifts),
        "temperature": round(random.uniform(25, 90), 2),
        "pressure": round(random.uniform(80, 150), 2),
        "cycle_time_sec": round(random.uniform(20, 90), 2),
        "output_qty": output_qty,
        "defect_qty": defect_qty,
        "machine_status": status
    }

    machine_logs.append(row)
```

```python
# Products that are ordered more frequently
popular_products = [
    "P001", "P002", "P003", "P004", "P005",
    "P006", "P007", "P008", "P009", "P010"
]

# Machine capacities
machine_capacity = {
    "MC001": 120,
    "MC002": 110,
    "MC003": 100,
    "MC004": 90,
    "MC005": 95,
    "MC006": 85,
    "MC007": 80,
    "MC008": 75,
    "MC009": 70,
    "MC010": 65
}

# Customers and their typical order quantity
customer_data = {
    "Bosch": (700, 1200),
    "Toyota": (600, 1000),
    "Tata Motors": (500, 900),
    "Mahindra": (400, 800),
    "Ashok Leyland": (350, 750),
    "TVS": (200, 600),
    "Hyundai": (450, 850),
    "Honda": (300, 700),
    "Volvo": (300, 700),
    "Bharat Benz": (350, 750)
}

order_status = [
    "Completed",
    "In Progress",
    "Scheduled",
    "Cancelled"
]
```

```python
orders = []

for i in range(1, 1001):

    # Popular products appear more often
    if random.random() < 0.60:
        product_id = random.choice(popular_products)
    else:
        product_id = random.choice(product_df["product_id"].tolist())

    product = product_df[product_df["product_id"] == product_id].iloc[0]

    # Customer
    customer = random.choice(list(customer_data.keys()))

    min_qty, max_qty = customer_data[customer]

    planned_qty = random.randint(min_qty, max_qty)

    # Machine
    machine = random.choice(list(machine_capacity.keys()))

    capacity = machine_capacity[machine]

    produced_qty = random.randint(
        int(planned_qty * 0.90),
        planned_qty
    )

    rejected_qty = round(
        produced_qty * random.uniform(0.005, 0.03)
    )

    efficiency = round(
        (produced_qty / planned_qty) * 100,
        2
    )

    orders.append({
        "order_id": f"PO{i:05}",
        "product_id": product["product_id"],
        "product_name": product["product_name"],
        "customer": customer,
        "machine_id": machine,
        "order_date": fake.date_between(
            start_date="-6M",
            end_date="today"
        ),
        "planned_quantity": planned_qty,
        "produced_quantity": produced_qty,
        "rejected_quantity": rejected_qty,
        "efficiency_percent": efficiency,
        "order_status": random.choices(
            order_status,
            weights=[65,20,10,5],
            k=1
        )[0]
    })
```

```python
production_orders = pd.DataFrame(orders)

production_orders.head()
```

```python
production_orders.to_csv(
    "../data/raw/mes/production_orders.csv",
    index=False
)

print("production_orders.csv created successfully")
print("Total Orders :", len(production_orders))
```

```python
#downtime_events.csv
machine_failure_weight = {
    "MC001": 3,
    "MC002": 5,
    "MC003": 6,
    "MC004": 8,
    "MC005": 4,
    "MC006": 9,
    "MC007": 7,
    "MC008": 3,
    "MC009": 2,
    "MC010": 1
}

reasons = [
    "Machine Breakdown",
    "Preventive Maintenance",
    "Power Failure",
    "Tool Change",
    "Quality Inspection",
    "Operator Unavailable",
    "Material Shortage",
    "Sensor Failure"
]

status = [
    "Resolved",
    "Pending",
    "In Progress"
]

technicians = [
    "Ramesh",
    "Suresh",
    "Anil",
    "Kiran",
    "Rahul",
    "Meena",
    "Priya",
    "Asha"
]
```

```python
downtime = []

for i in range(1, 301):

    machine = random.choices(
        list(machine_failure_weight.keys()),
        weights=machine_failure_weight.values(),
        k=1
    )[0]

    duration = random.randint(10, 180)

    downtime.append({
        "event_id": f"DT{i:04}",
        "machine_id": machine,
        "event_date": fake.date_between(
            start_date="-6M",
            end_date="today"
        ),
        "reason": random.choice(reasons),
        "downtime_minutes": duration,
        "technician": random.choice(technicians),
        "status": random.choice(status)
    })
```

```python
downtime_df = pd.DataFrame(downtime)

downtime_df.head()
downtime_df.to_csv(
    "../data/raw/mes/downtime_events.csv",
    index=False
)

print("downtime_events.csv created successfully")
print("Total Records :", len(downtime_df))
```
